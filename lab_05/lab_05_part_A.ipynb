## linear regression and logistic regression models using scikit learn

```python
from sklearn.model_selection import train_test_split

indices = df.index

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.2,
    random_state=42
)

train_df = df.loc[train_idx]
test_df = df.loc[test_idx]
```

```python
train_df.to_csv("train_data.csv", index=False)
test_df.to_csv("test_data.csv", index=False)
```

```python
X_reg_train = train_df.drop(
    columns=["actual_productivity", "MeetsTarget"]
)

y_reg_train = train_df["actual_productivity"]

X_reg_test = test_df.drop(
    columns=["actual_productivity", "MeetsTarget"]
)

y_reg_test = test_df["actual_productivity"]
```

```python
X_clf_train = train_df.drop(
    columns=["actual_productivity", "MeetsTarget"]
)

y_clf_train = train_df["MeetsTarget"]

X_clf_test = test_df.drop(
    columns=["actual_productivity", "MeetsTarget"]
)

y_clf_test = test_df["MeetsTarget"]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

numerical_features = [
    "targeted_productivity",
    "smv",
    "wip",
    "overtime_per_worker",
    "incentive",
    "idle_time",
    "idle_men",
    "no_of_style_change",
    "no_of_workers"
]

categorical_features = [
    "quarter",
    "department",
    "day",
    "team"
]


numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])



preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numerical_features),
    ("cat", categorical_pipeline, categorical_features)
])
```

```python
from sklearn.linear_model import LinearRegression
import time

regression_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])


# Training time
start_time = time.perf_counter()
regression_model.fit(X_reg_train, y_reg_train)
train_time = time.perf_counter() - start_time


# Prediction time
start_time = time.perf_counter()
y_reg_pred = regression_model.predict(X_reg_test)
prediction_time = time.perf_counter() - start_time


print("Training time   :", train_time, "seconds")
print("Prediction time :", prediction_time, "seconds")
```

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
import numpy as np

mae = mean_absolute_error(y_reg_test, y_reg_pred)

mse = mean_squared_error(y_reg_test, y_reg_pred)
rmse = np.sqrt(mse)

r2 = r2_score(y_reg_test, y_reg_pred)

print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
from sklearn.linear_model import LogisticRegression
import time

classification_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])


start_time = time.perf_counter()
classification_model.fit(X_clf_train, y_clf_train)
clf_train_time = time.perf_counter() - start_time


# Prediction time
start_time = time.perf_counter()
y_clf_pred = classification_model.predict(X_clf_test)
clf_prediction_time = time.perf_counter() - start_time


print("Training time   :", clf_train_time, "seconds")
print("Prediction time :", clf_prediction_time, "seconds")
```

```python
y_clf_prob = classification_model.predict_proba(X_clf_test)

print(y_clf_prob[:5])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_clf_test, y_clf_pred)
precision = precision_score(y_clf_test, y_clf_pred)
recall = recall_score(y_clf_test, y_clf_pred)
f1 = f1_score(y_clf_test, y_clf_pred)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1-score :", f1)
```
