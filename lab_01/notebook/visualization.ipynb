```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud

import re

plt.style.use("ggplot")
sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("../data/books_clean.csv")
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["price"], bins=20, kde=True)

plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")

plt.tight_layout()
plt.savefig("../plots/price_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="rating")

plt.title("Distribution of Book Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Books")

plt.tight_layout()
plt.savefig("../plots/rating_distribution.png", dpi=300)

plt.show()
```

```python
avg_price = (
    df.groupby("category")["price"]
      .mean()
      .sort_values()
)

plt.figure(figsize=(10,7))

avg_price.plot(kind="barh")

plt.title("Average Book Price by Category")
plt.xlabel("Average Price (£)")
plt.ylabel("Category")

plt.tight_layout()
plt.savefig("../plots/average_price_by_category.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="price",
    y="rating",
    alpha=0.7
)

plt.title("Price vs Rating")
plt.xlabel("Price (£)")
plt.ylabel("Rating")

plt.tight_layout()
plt.savefig("../plots/price_vs_rating.png", dpi=300)

plt.show()
```

```python
text = " ".join(df["product_description"])

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color="white"
).generate(text)

plt.figure(figsize=(14,7))

plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")

plt.title("Word Cloud of Book Descriptions")

plt.tight_layout()
plt.savefig("../plots/wordcloud.png", dpi=300)

plt.show()
```

```python
top_value = (
    df.nlargest(10, "value_score")
      .sort_values("value_score")
)

plt.figure(figsize=(10,6))

plt.barh(
    top_value["title"],
    top_value["value_score"]
)

plt.title("Top 10 Books by Value Score")
plt.xlabel("Value Score")
plt.ylabel("Book")

plt.tight_layout()
plt.savefig("../plots/top_value_books.png", dpi=300)

plt.show()
```
