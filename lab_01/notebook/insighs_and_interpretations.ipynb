```python
import pandas as pd
df = pd.read_csv("../data/books_clean.csv")
```

```python

print("INSIGHTS AND INTERPRETATION")

print(f"\n1. A total of {len(df)} books were analyzed across {df['category'].nunique()} different categories.")

most_common_category = df["category"].value_counts().idxmax()
most_common_count = df["category"].value_counts().max()

print(
    f"\n2. The most represented category is '{most_common_category}' "
    f"with {most_common_count} books."
)

highest_price_category = (
    df.groupby("category")["price"]
      .mean()
      .idxmax()
)

highest_price = (
    df.groupby("category")["price"]
      .mean()
      .max()
)

print(
    f"\n3. '{highest_price_category}' has the highest average book price "
    f"at £{highest_price:.2f}."
)

highest_rating_category = (
    df.groupby("category")["rating"]
      .mean()
      .idxmax()
)

highest_rating = (
    df.groupby("category")["rating"]
      .mean()
      .max()
)

print(
    f"\n4. '{highest_rating_category}' has the highest average rating "
    f"of {highest_rating:.2f}/5."
)

correlation = df["price"].corr(df["rating"])

print(
    f"\n5. The correlation between price and rating is {correlation:.3f}, "
    "indicating "
    + (
        "a weak relationship."
        if abs(correlation) < 0.3
        else "a moderate relationship."
    )
)

best_book = df.loc[df["value_score"].idxmax()]

print(
    f"\n6. The book with the highest value score is "
    f"'{best_book['title']}' "
    f"(Rating: {best_book['rating']}/5, "
    f"Price: £{best_book['price']:.2f}, "
    f"Value Score: {best_book['value_score']:.3f})."
)

print(
    f"\n7. The average stock available per book is "
    f"{df['stock'].mean():.1f} copies."
)

print("LIMITATIONS")


print("""
• Only the first 200 books (10 catalogue pages) were scraped.
• The website is intended for web scraping practice, so prices and ratings
  may not represent a real online bookstore.
• Some product descriptions in the source HTML contain duplicated or
  truncated text, which may affect text-based analysis.
• The website does not provide customer reviews, so analysis is limited
  to product descriptions and metadata.
""")
```
