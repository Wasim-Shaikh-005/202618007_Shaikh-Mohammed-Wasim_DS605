```python
A = np.array([[1,2,3],[4,5,6],[7,8,10]])
B = np.array([[2,1,0],[1,3,2],[4,0,1]])

print("Addition:\n", A+B)
print("\nElement-wise multiplication:\n", A*B)
print("\nMatrix multiplication:\n", A @ B)

AT = A.T
det = np.linalg.det(A)

print("\nTranspose:\n", AT)
print("Determinant:", det)

if det != 0:
    inv = np.linalg.inv(A)
    print("\nInverse:\n", inv)
    print("\nVerification:", np.allclose(A @ inv, np.eye(3)))
```

## task 5

```python
male_50 = df[(df.Sex=="male") & (df.Age>50)]
print(" Male passengers older than 50:", len(male_50))
```

```python
female_first = df[(df.Sex=="female") & (df.Pclass==1)]
rate = female_first.Survived.mean()*100
print(" Female first class:", len(female_first))
print("   Survival %:", round(rate,2))
```

```python
median_fare = df.Fare.median()
group = df[(df.Age.between(20,40)) & (df.Fare>median_fare) & (df.Survived==1)]
print("Age 20-40, Fare above median, survived:", len(group))
```

```python
alone = df[(df.SibSp==0) & (df.Parch==0) & (df.Age<30) & (df.Survived==0)]
print(" Travelling alone, age<30, did not survive:", len(alone))
```

```python
south_median = df[df.Embarked=="S"].Fare.median()
result = df[(df.Embarked=="S") & (df.Pclass.isin([2,3])) & (df.Fare>south_median)]
print(" Embarked=S, Pclass 2/3, Fare above Southampton median:", len(result))
```

## task 6

```python
print("Survival rate by Sex")
print(df.groupby("Sex")["Survived"].mean())

print("\nSurvival rate by Pclass")
print(df.groupby("Pclass")["Survived"].mean())

print("\nAverage Age and Fare by Pclass")
print(df.groupby("Pclass")[["Age","Fare"]].mean())

print("\nPassenger count and survival rate by Sex-Pclass")
print(df.groupby(["Sex","Pclass"]).agg(
    PassengerCount=("PassengerId","count"),
    SurvivalRate=("Survived","mean")
))

print("\nPassenger count, average Fare and survival rate by Embarked")
print(df.groupby("Embarked").agg(
    PassengerCount=("PassengerId","count"),
    AverageFare=("Fare","mean"),
    SurvivalRate=("Survived","mean")
))
```

## task 7

```python
missing = df.isnull().sum()
percent = (missing/len(df))*100

summary = pd.DataFrame({
    "Missing": missing,
    "Percent": percent.round(2)
})
print(summary)
```

```python
missing.plot(kind="bar")
plt.title("Missing Values")
plt.ylabel("Count")
plt.show()
```

```python
print("Before filling Age:")
print(df["Age"].isnull().sum())
```

```python
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())
print("After filling with mean:")
print(df_mean["Age"].isnull().sum())
```

```python
df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

df_mode = df.copy()
df_mode["Age"] = df_mode["Age"].fillna(df_mode["Age"].mode()[0])

df_random = df.copy()
na = df_random["Age"].isna()
random_values = np.random.choice(df_random["Age"].dropna(), na.sum())
df_random.loc[na, "Age"] = random_values

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR

outliers = df[(df.Fare<lower) | (df.Fare>upper)]
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Outlier Count:", len(outliers))
```

## Observations

1. Female passengers had a much higher survival rate than male passengers.
2. First-class passengers survived more often than second and third class.
3. Fare has a positive relationship with survival because higher fares usually belong to higher classes.
4. Pclass has a negative correlation with survival since larger class numbers represent lower classes.
5. Most missing values are present in the Age and Cabin columns.
6. Fare contains several high-value outliers based on the IQR method.
7. Passengers travelling alone generally had lower survival compared to those with family.
